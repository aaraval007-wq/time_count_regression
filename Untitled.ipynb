```python
import numpy as np
import pandas as pd

from count_regression import CountRegressor
```

```python
%load_ext autoreload
%autoreload 2
```

```python
data = pd.read_csv("football_data_adjcols.csv")

X_home = data.drop(columns = ["home_shots","away_shots"])
X_away = data.drop(columns = ["away_shots", "home_shots"])

y_home = data["home_shots"]
y_away = data["away_shots"]
```

```python
home_regression = CountRegressor(X_home, y_home, 1000, 10)
```
